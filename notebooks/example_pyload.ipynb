```python
import os
os.chdir('/content/drive/MyDrive/credit-risk-assessment-system')

import pandas as pd
import json

X_test = pd.read_csv('data/processed/X_test_fe.csv')

leftover_cols = [c for c in ['SK_ID_CURR', 'AMT_ANNUITY_RAW', 'AMT_CREDIT_RAW', 'AMT_INCOME_TOTAL_RAW',
                               'CODE_GENDER_M', 'CODE_GENDER_XNA']
                  if c in X_test.columns]
X_test_fair = X_test.drop(columns=leftover_cols)

print(X_test_fair.shape)
```

```python
example_payload = {"features": X_test_fair.iloc[0].to_dict()}

print(json.dumps(example_payload, indent=2)[:1000])

with open('reports/example_api_payload.json', 'w') as f:
    json.dump(example_payload, f, indent=2)

print("Saved payload")
```

```python
raw_df = pd.read_csv('data/raw/application_train.csv')
X_raw = raw_df.drop(columns=['TARGET'])
y_raw = raw_df['TARGET']

from sklearn.model_selection import train_test_split
_, X_test_raw, _, _ = train_test_split(X_raw, y_raw, test_size=0.2, random_state=42, stratify=y_raw)

scoring_results = pd.read_csv('reports/risk_scoring_results.csv')

fairness_df = scoring_results.copy()
fairness_df['gender'] = X_test_raw['CODE_GENDER'].values

fairness_df.to_csv('reports/fairness_analysis.csv', index=False)
print(fairness_df['gender'].value_counts())
print("Saved fairness_analysis.csv")
```
